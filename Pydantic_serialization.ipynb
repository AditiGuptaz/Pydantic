```python
from pydantic import BaseModel
```

```python
#Serialization in Pydantic means converting a model object into a plain format like a dictionary or JSON string — 
#so it can be sent over a network, saved to a file, or passed to another system.

#Serialization = turning a Pydantic model (a Python object) into a simple, transferable 
#format (a dict or a JSON string).
```

```python
class Address(BaseModel):
    city:str
    state:str
    pin:str 

class Patient(BaseModel):
    name:str
    gender:str = 'Male'
    age:int 
    address: Address  
```

```python
address_dict ={'city':'somerset', 'state': 'New Jersey', 'pin':'08873'}
address1=Address(**address_dict)
```

```python
patient_dict = {'name':'Aditi',  'age':30,'address':address1}  #'gender':'Female', #here, gender was not set during the object creation so it will not export.
patient1 = Patient(**patient_dict)
```

```python
#temp = patient1.model_dump()  #for python dict

#temp = patient1.model_dump(include=['name','gender']) #include parameter if want to export only name

#temp = patient1.model_dump(exclude=['name','gender'])  #exclude name , gender

#temp = patient1.model_dump(exclude= {'address':['state']})   #if want to exclude only the state from address

#exclude_unset
temp = patient1.model_dump(exclude_unset = False)
```

```python
print(temp)
print(type(temp))
print(patient1)
print(type(patient1))
```
