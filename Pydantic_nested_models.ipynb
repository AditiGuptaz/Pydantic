```python
from pydantic import BaseModel
```

```python
#Nested Model definition
#Nested models in Pydantic means using one model as a field inside another model.
#so you can represent structured data that has structure within it.. 
```

```python
class Address(BaseModel):
    city:str
    state:str
    pin:str 

class Patient(BaseModel):
    name:str
    gender:str
    age:int 
    address: Address  
```

```python
address_dict ={'city':'somerset', 'state': 'New Jersey', 'pin':'08873'}
address1=Address(**address_dict)
```

```python
patient_dict = {'name':'Aditi', 'gender':'Female', 'age':30,'address':address1}
patient1 = Patient(**patient_dict)
```

```python
print(patient1)
```

```python
print(patient1.name)
print(patient1.address.city)
print(patient1.address.pin)
```

```python
#Benefits of using Nested models
#1. Better organization of related data (eg. vitals, address, insurance)
#2. Reusability : use vitals in multiple models (eg: Patient, Medical record)
#3. Readability : Easier for developers and API consumers to understand.
#4. Validation : Nested models are validated automatically - No extra work needed.
```
